```python
from box import ConfigBox
```

```python
d2 = ConfigBox({"key":"value" , "key1":"value1"})
```

```python
d2
```

```python
#as we see below, usage of ConfigBox in our dictionary enables us to access a value even by writing 'd2.key'
d2.key
#for the ease of using configurations from the yaml file during the whole project, we use "ConfigBox"
```

```python
def get_prodict(x:int, y:int) -> int:
    return x * y
```

```python
get_prodict(x=2, y=4)
```

```python
get_prodict(x=2, y="4")
#this happens only because ensure annotation is not applied over here.
```

```python
from ensure import ensure_annotations
#this ensures that code which doesn't check its datatype constraint does not run at all and gives error.
```

```python
@ensure_annotations
def get_product(x: int, y: int):
    return x * y

print(get_product.__annotations__)

get_product(2, "4")
```
